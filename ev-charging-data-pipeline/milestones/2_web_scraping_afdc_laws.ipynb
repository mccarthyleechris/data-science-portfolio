Performing 5 Data Transformations on Website

https://afdc.energy.gov/laws/matrix?sort_by=tech

Load html file and show head / tail

```python
import pandas as pd

pd.set_option("display.max_columns", None)

url = "https://afdc.energy.gov/laws/matrix?sort_by=tech"

tables = pd.read_html(url)
tech_df = tables[0]

print("Tech/Fuel table shape:", tech_df.shape)
display(tech_df.head())
display(tech_df.tail())
```

Data Transformation 1 - Standardize Column Names

```python
tech_df.columns = (
    tech_df.columns
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("(", "", regex=False)
    .str.replace(")", "", regex=False)
)

display(tech_df.head())
```

Similar to milestone 2, I went ahead and standardized the casing of each column. Everything is lowercase and spaces have been replaced with underscores. This is to reflect a database environment.

Data Transformation 2 - Remove Totals Row

```python
rows_before = tech_df.shape[0]

tech_df = tech_df[tech_df["jurisdiction"] != "Totals"]

rows_after = tech_df.shape[0]

print("Rows removed:", rows_before - rows_after)
display(tech_df.head())
```

Initially, there is a 'Totals' row for each category of technology in correspondence to jurisdiction. While I can see this being useful in some regard, I think removing it to better focus on differences between jurisdictions will tell a better story. I'd like to see just how different state to state is. And how much support the fed on its own is putting forth. This also gives me room to bucket things by region if I so choose. Currently, my focus is WA state but I have options now.

Data Transformation 3 - Reshape to Long Format

```python
rows_before = tech_df.shape[0]

tech_long = tech_df.melt(
    id_vars="jurisdiction",
    var_name="technology",
    value_name="count"
)

rows_after = tech_long.shape[0]

print("Rows before:", rows_before)
print("Rows after:", rows_after)
display(tech_long.head(15))
```

I had to do some digging to understand why this transformation is important. After further research, I learned a couple things. While the wide view is good for display (which I liked about it), the long view is better for analysis because of how pandas handles single variable measurements. Ultimately, it seems to be the case that it's commonplace to swap back and forth between the two styles of display. I think this is a good opportunity to practice.

Data Transformation 4 - Remove Zero Counts

```python
rows_before = tech_long.shape[0]

tech_long = tech_long[tech_long["count"] > 0]

rows_after = tech_long.shape[0]

print("Rows removed:", rows_before - rows_after)
display(tech_long.head(15))
```

I'm particularly a fan of this specific transformation in this scenario. My reasoning is because it helps declutter the long format list. By removing rows where no policy exists for the technology/jurisdiction pairing, we clean up the dataset while maintaining the importance of the values in it.

Data Transformation 5 - Matching Technology Labels

```python
tech_long["technology"] = tech_long["technology"].str.replace("_", " ", regex=False).str.title()

display(tech_long.head(15))
```

Different from the column standardization (where we replaced spaces and uppercase with underscores and lowercase), changing the labels in the technology column to match the casing and spacing of the other values makes sense. The jurisdictions are capitalized so having technology look the same way helps with legibility and flow.

Data Transformation 6 - Outliers

```python
#running IQR

Q1 = tech_long["count"].quantile(0.25)
Q3 = tech_long["count"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
#flagging outliers

tech_long["outlier"] = (
    (tech_long["count"] < lower_bound) |
    (tech_long["count"] > upper_bound)
)
```

```python
#counting and previewing outliers

outlier_count = tech_long["outlier"].sum()
print("Number of outliers detected:", outlier_count)

tech_long[tech_long["outlier"]].head(15)
```

Similar to milestone 2, I went ahead and tested for outliers using IQR. Instead of doing anything permanent, I flagged them and counted them so I could analyze. Current educated guesses lead me to believe that outliers regarding EVs have to do with density and popularity in relation to jurisdiction. For example, California has a count of 145 EV policies which is by and large an outlier but relevant to the story specific to California itself. 

On a side note, I notice here that while most 'technology' labels are improved by my previous transformation (biodiesel -> Biodiesel), anything that is an acronym looks a bit weird (evs is Evs, propane lpg is Propane Lpg). While it's still an improvement from all lowercase, I want to find a way to properly capitalize acronyms for future analysis.
